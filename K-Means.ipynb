## 3. Estandarización

`StandardScaler` para que las tres variables contribuyan equitativamente.

```python
X = df[['edad', 'horas_estudio', 'promedio_academico']].values
escalador = StandardScaler()
X_esc = escalador.fit_transform(X)
```

## 4. Método del codo (paleta púrpura)

```python
inercias = []
valores_k = range(1, 11)

for k in valores_k:
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X_esc)
    inercias.append(km.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(valores_k, inercias, marker='D', color='#5e35b1',
         linewidth=2.5, markersize=8, markerfacecolor='#ff6f61',
         markeredgecolor='#2d2d3a')
plt.title('Método del Codo — Elección de K', color='#5e35b1', fontsize=13)
plt.xlabel('Cantidad de Clusters (K)')
plt.ylabel('Inercia')
plt.xticks(valores_k)
plt.grid(True, alpha=0.25, linestyle='--')
plt.savefig('codo_estudiantes.png', dpi=150, bbox_inches='tight')
plt.show()
```

La curva se estabiliza a partir de **K = 4** (y la consigna pide 4). Ese será el valor elegido.

## 5. Modelo final con K = 4

```python
modelo = KMeans(n_clusters=4, random_state=42)
df['cluster'] = modelo.fit_predict(X_esc)
print('Inercia final:', modelo.inertia_)
print('\nEstudiantes por cluster:')
print(df.groupby('cluster').size())
df.sort_values('cluster').head(10)
```

## 6. Visualización con paleta púrpura / coral / turquesa / amarillo

```python
paleta = {0: '#5e35b1', 1: '#ff6f61', 2: '#26c6da', 3: '#f4b400'}

plt.figure(figsize=(10, 6))
for c in range(4):
    grupo = df[df['cluster'] == c]
    plt.scatter(grupo['edad'], grupo['promedio_academico'],
                label=f'Grupo {c}', color=paleta[c],
                s=90, edgecolor='#2d2d3a', linewidth=0.6, alpha=0.9)

plt.xlabel('Edad')
plt.ylabel('Promedio académico')
plt.title('Segmentación de Estudiantes — K-Means', color='#5e35b1', fontsize=13)
plt.legend(frameon=True, edgecolor='#5e35b1')
plt.grid(True, alpha=0.25, linestyle='--')
plt.tight_layout()
plt.savefig('grafico_estudiantes.png', dpi=150, bbox_inches='tight')
plt.show()
```

## 7. Interpretación pedagógica

- 🟣 **Grupo 0 (púrpura):** estudiantes jóvenes con muchas horas de estudio y promedios bajos → *principiantes esforzados*.
- 🟠 **Grupo 1 (coral):** estudiantes de mediana edad con horas moderadas y promedios medios → *perfil estable*.
- 🔵 **Grupo 2 (turquesa):** estudiantes mayores con pocas horas de estudio y altos promedios → *experimentados*.
- 🟡 **Grupo 3 (amarillo):** estudiantes jóvenes con pocas horas de estudio y bajos promedios → *desmotivados*.

La inercia final con K=4 es la reportada en la celda anterior, indicando una cohesión razonable.
